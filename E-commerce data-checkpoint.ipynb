```python
import pandas as pd

df = pd.read_csv(r"D:\Data Analytics\e commerce dataset formet.csv",encoding = 'latin-1')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
## convert Discount_Percentage
df['Discount_Percentage'] = df['Discount_Percentage'].astype(float)
```

```python
df['Discount_Percentage']
```

```python
## KPI METRICS

Total_sales = df['Total sales'].sum()
Total_orders = df['Order_ID'].nunique()
avg_order_value = df['Total sales'].mean()
avg_session_time = df['Session_Duration_Minutes'].mean()
avg_rating = df['Customer_Rating'].mean()

print("Total Sales:", Total_sales)
print("Total Orders:", Total_orders)
print("Average Order Value:", avg_order_value)
print("Average Session Duration:", avg_session_time)
print("Average Customer Rating:",avg_rating )
```

```python
##Filter by Product Category
electronics_df = df[df['Product_Category'] == 'Electronics']
electronics_df
```

```python
##Filter by City
Bursa_df = df[df['City'] =='Bursa']
Bursa_df
```

```python
##Filter by Payment Method
credit_card_df = df[df['Payment_Method'] == 'Credit Card']
credit_card_df
```

```python
##KPI AFTER FILTER
electronics_revenue = electronics_df['Total sales'].sum()
electronics_orders = electronics_df['Order_ID'].nunique()

print("Electronics Revenue:", electronics_revenue)
print("Electronics Orders:", electronics_orders)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
sales_by_date = df.groupby('Date')['Total sales'].sum()
sales_by_date.plot(kind='line', title='Sales Trend')
plt.show()
```

```python
category_sales = df.groupby('Product_Category')['Total sales'].sum()
category_sales.plot(kind='bar', title='Revenue by Category')
plt.show()
```

```python
plt.scatter(df['Session_Duration_Minutes'], df['Total sales'])
plt.xlabel('Session Duration')
plt.ylabel('Total Sales')
plt.title('Session Duration vs Sales')
plt.show()
```
